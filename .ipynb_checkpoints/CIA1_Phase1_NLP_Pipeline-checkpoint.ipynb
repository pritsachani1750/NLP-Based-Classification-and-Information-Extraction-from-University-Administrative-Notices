# CIA 1 – Phase 1 NLP Pipeline
## NLP-Based Classification and Information Extraction from University Administrative Documents

**Dataset source:** University of Hyderabad official notices archive: https://www.uohyd.ac.in/or/all-notices/

This notebook demonstrates the Phase 1 workflow: dataset loading, text normalization, tokenization, stop-word removal, stemming, lemmatization, regex candidate extraction, Bag-of-Words and TF-IDF.

**Scope note:** This is a preprocessing and feature-extraction pipeline. It does not train a supervised classifier; classification is planned for Phase 2. OCR output is machine-generated and must be reviewed before being treated as verified text.

```python

from pathlib import Path
import re
import unicodedata
import warnings

import pandas as pd
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Project root
PROJECT_ROOT = Path.cwd()

# Supported dataset filenames
CANDIDATE_NAMES = [
    "uoh_phase1_preprocessed_improved.csv",
    "uoh_phase1_preprocessed_final.csv",
    "uoh_phase1_dataset_with_ocr_candidates.csv",
    "uoh_admin_documents_cleaned_annotation_prep.csv",
]

# Find matching datasets recursively
matches = []

for name in CANDIDATE_NAMES:
    matches.extend(PROJECT_ROOT.rglob(name))

if not matches:
    raise FileNotFoundError(
        "No supported dataset found. Check your project's data folder."
    )

# Prefer the data folder, then the improved dataset
matches.sort(
    key=lambda p: (
        0 if p.parent.name.lower() == "data" else 1,
        CANDIDATE_NAMES.index(p.name),
        len(p.parts)
    )
)

DATASET_PATH = matches[0]

print("Dataset selected:", DATASET_PATH)

# Load dataset
df = pd.read_csv(DATASET_PATH)

print("Total records:", len(df))
print("Dataset columns:", len(df.columns))
print("Available columns:", df.columns.tolist())
```

```python
df = pd.read_csv(DATASET_PATH, low_memory=False)
print("Total records:", len(df))
print("Columns:", len(df.columns))
print("Available text-related columns:", [c for c in df.columns if "text" in c.lower()][:20])

# Prefer the NLP source text, then notice_text, then cleaned_text.
text_col = next((c for c in ["original_text_for_nlp", "notice_text", "cleaned_text", "processed_text"] if c in df.columns), None)
if text_col is None:
    raise ValueError("No suitable text column was found.")
raw = df[text_col].fillna("").astype(str)
print("Selected text column:", text_col)
print("Records with text:", int(raw.str.strip().ne("").sum()))
print("Records with empty text:", int(raw.str.strip().eq("").sum()))
```

```python
def normalize_text(text):
    text = unicodedata.normalize("NFKC", text)
    text = re.sub(r"[\\x00-\\x08\\x0B\\x0C\\x0E-\\x1F]", " ", text)
    text = re.sub(r"\\s+", " ", text)
    return text.strip()

df["normalized_text_demo"] = raw.map(normalize_text)
sample_idx = df["normalized_text_demo"].str.len().idxmax()
sample_raw = raw.loc[sample_idx]
sample_normalized = df.loc[sample_idx, "normalized_text_demo"]
print("Sample record:", df.loc[sample_idx, "notice_id"] if "notice_id" in df.columns else sample_idx)
print("\nRAW TEXT (first 700 chars):\n", sample_raw[:700])
print("\nNORMALIZED TEXT (first 700 chars):\n", sample_normalized[:700])
```

```python
# Download required NLTK resources if they are not already available.
for resource in ["punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"]:
    try:
        nltk.download(resource, quiet=True)
    except Exception as exc:
        warnings.warn(str(exc))

stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def process_document(text):
    if not text:
        return [], [], []
    try:
        tokens = word_tokenize(text.lower())
    except LookupError:
        tokens = re.findall(r"\\b\\w+\\b", text.lower())
    filtered = [t for t in tokens if t.isalpha() and t not in stop_words]
    stemmed = [stemmer.stem(t) for t in filtered]
    lemmatized = [lemmatizer.lemmatize(t) for t in filtered]
    return tokens, stemmed, lemmatized

sample_tokens, sample_stems, sample_lemmas = process_document(sample_normalized)
print("Tokens:", sample_tokens[:35])
print("\nAfter stemming:", sample_stems[:35])
print("\nAfter lemmatization:", sample_lemmas[:35])
print("\nToken count:", len(sample_tokens))
print("Filtered token count:", len(sample_lemmas))
```

```python
# Regular expressions: dates, email addresses, and labelled reference numbers.
date_pattern = r"\\b(?:\\d{1,2}[/-]\\d{1,2}[/-]\\d{2,4}|\\d{1,2}\\s+(?:January|February|March|April|May|June|July|August|September|October|November|December)\\s+\\d{4}|(?:January|February|March|April|May|June|July|August|September|October|November|December)\\s+\\d{1,2},?\\s+\\d{4})\\b"
email_pattern = r"\\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\\.[A-Za-z]{2,}\\b"
reference_pattern = r"\\b(?:Ref(?:erence)?\\s*No\\.?|File\\s*No\\.?|Notification\\s*No\\.?|Tender\\s*No\\.?|NIT\\s*No\\.?|Letter\\s*No\\.?|Proceedings\\s*No\\.?)\\s*[:#-]?\\s*[A-Z0-9][A-Z0-9/_-]*"

sample_dates = re.findall(date_pattern, sample_normalized, flags=re.IGNORECASE)
sample_emails = re.findall(email_pattern, sample_normalized, flags=re.IGNORECASE)
sample_refs = re.findall(reference_pattern, sample_normalized, flags=re.IGNORECASE)
print("Date candidates:", sample_dates[:15])
print("Email candidates:", sample_emails[:15])
print("Reference candidates:", sample_refs[:15])
print("\nReminder: regex matches are candidates and need contextual verification.")
```

```python
# Apply the text-processing steps to the complete dataset.
processed = []
for text in df["normalized_text_demo"]:
    _, _, lemmas = process_document(text)
    processed.append(" ".join(lemmas))

df["processed_text_demo"] = processed
usable = df["processed_text_demo"].str.strip().ne("")
docs = df.loc[usable, "processed_text_demo"].tolist()
print("Total records:", len(df))
print("Documents used for vectorization:", len(docs))
print("Documents with no processed text:", int((~usable).sum()))
```

```python
bow = CountVectorizer(max_features=3000, min_df=1)
tfidf = TfidfVectorizer(max_features=3000, min_df=1)

bow_matrix = bow.fit_transform(docs)
tfidf_matrix = tfidf.fit_transform(docs)

print("Bag-of-Words matrix shape:", bow_matrix.shape)
print("TF-IDF matrix shape:", tfidf_matrix.shape)

bow_vocab = pd.DataFrame({"term": bow.get_feature_names_out()})
tfidf_vocab = pd.DataFrame({"term": tfidf.get_feature_names_out()})
display(bow_vocab.head(15))
display(tfidf_vocab.head(15))
```

```python
# Save notebook outputs and reproducible artifacts.
OUTPUT_DIR = PROJECT_ROOT / "cia1_notebook_outputs"
OUTPUT_DIR.mkdir(exist_ok=True)

df.to_csv(OUTPUT_DIR / "uoh_phase1_notebook_processed.csv", index=False)
bow_vocab.to_csv(OUTPUT_DIR / "bow_vocabulary.csv", index=False)
tfidf_vocab.to_csv(OUTPUT_DIR / "tfidf_vocabulary.csv", index=False)

from scipy.sparse import save_npz
save_npz(OUTPUT_DIR / "bow_matrix.npz", bow_matrix)
save_npz(OUTPUT_DIR / "tfidf_matrix.npz", tfidf_matrix)

print("Saved outputs to:", OUTPUT_DIR.resolve())
print("Classifier training: not performed in CIA 1; planned for Phase 2.")
```

## Results and limitations

- Bag-of-Words and TF-IDF provide numerical representations of the processed documents.
- Empty-text records are retained in the dataset but excluded from vectorization.
- Date, email and reference-number regex matches are candidate extractions, not validated facts.
- English NLTK preprocessing is a baseline and may need language-specific handling for non-English documents.
- OCR-derived text requires manual quality review.
- No supervised classifier is trained in this Phase 1 notebook.
